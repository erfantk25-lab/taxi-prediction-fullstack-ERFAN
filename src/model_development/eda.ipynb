## EDA och data cleaning

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/taxi_trip_pricing.csv")

df.head()
```

## Examine metadata

```python
df.info()
```

## Describe the data

```python
df.describe()
```

## Visualize

```python
df['Trip_Price'].hist(bins=50)
plt.xlabel("Pris")
plt.ylabel("Antal resor")
plt.title("Fördelning av taxipriser")
plt.show()

```

## Handling null values

```python

df.isnull().sum()

df_missing_label = df[df['Trip_Price'].isnull()]
df_missing_label.to_csv("../data/user_inputs.csv", index=False)

df = df.dropna(subset=['Trip_Price'])
```

## Remove outliers (unreasonable prices)

```python
df = df[(df['Trip_Price'] > 0) & (df['Trip_Price'] < 500)]
```

## Export Cleaned Data

```python
df.to_csv("../data/taxi_trip_clean.csv", index=False)
```
